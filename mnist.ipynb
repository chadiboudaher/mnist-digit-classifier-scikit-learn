# MNIST Dataset

this dataset include images of digits handwritten  by high school students and employees.


## Load The dataset

This data include 70000 images and each image has 784 features.

```python
from sklearn.datasets import fetch_openml

mnist = fetch_openml("mnist_784")
```

each image is 28x28 pixels = 784

```python
X, y = mnist["data"], mnist["target"]
X.shape, y.shape
```

```python
import numpy as np
X = np.array(X)
# X[36000]
```

```python
%matplotlib inline
import matplotlib
import matplotlib.pyplot as plt

# Get a random data sample
some_digit = X[3]

# Reshape the image
some_digit_image = some_digit.reshape(28, 28)

# Display the image of the chosen data sample
plt.imshow(some_digit_image, cmap=matplotlib.cm.binary)
plt.axis("off")
plt.show()
```
